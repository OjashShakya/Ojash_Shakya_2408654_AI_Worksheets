3. Volume:

– Convert liters (L) to gallons (gal).

– Convert gallons (gal) to liters (L).

```python
# Unit Converter Program

def convert_units(conversion_type, value):
    """
    Converts a value between different measurement units.

    Parameters:
    conversion_type (str): Type of conversion (m_to_ft, ft_to_m, kg_to_lb,
                           lb_to_kg, l_to_gal, gal_to_l).
    value (float): The value that needs to be converted.

    Returns:
    float: Converted value based on the selected conversion type.
    """

    if conversion_type == "m_to_ft":
        return value * 3.28084
    elif conversion_type == "ft_to_m":
        return value / 3.28084
    elif conversion_type == "kg_to_lb":
        return value * 2.20462
    elif conversion_type == "lb_to_kg":
        return value / 2.20462
    elif conversion_type == "l_to_gal":
        return value * 0.264172
    elif conversion_type == "gal_to_l":
        return value / 0.264172

print("\nLength Conversion")
print("1. Meters (m) to Feet (ft)")
print("2. Feet (ft) to Meters (m)")

print("\nWeight Conversion")
print("3. Kilograms (kg) to Pounds (lbs)")
print("4. Pounds (lbs) to Kilograms (kg)")

print("\nVolume Conversion")
print("5. Liters (L) to Gallons (gal)")
print("6. Gallons (gal) to Liters (L)")

try:
    choice = int(input("\nEnter your choice (1-6): "))
    value = float(input("Enter the value to convert: "))

    if choice == 1:
        result = convert_units("m_to_ft", value)
    elif choice == 2:
        result = convert_units("ft_to_m", value)
    elif choice == 3:
        result = convert_units("kg_to_lb", value)
    elif choice == 4:
        result = convert_units("lb_to_kg", value)
    elif choice == 5:
        result = convert_units("l_to_gal", value)
    elif choice == 6:
        result = convert_units("gal_to_l", value)
    else:
        print("Error: Unsupported conversion type.")
        result = None

    if result is not None:
        print("Converted value:", result)

except ValueError:
    print("Error: Invalid input. Please enter numeric values.")
```

```python
# Program to perform mathematical operations on a list of numbers

def find_sum(numbers):
    """
    Calculate the sum of numbers in a list.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The sum of all numbers in the list.
    """
    return sum(numbers)


def find_average(numbers):
    """
    Calculate the average of numbers in a list.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The average of the numbers in the list.
    """
    return sum(numbers) / len(numbers)


def find_maximum(numbers):
    """
    Find the maximum number in a list.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The largest number in the list.
    """
    return max(numbers)


def find_minimum(numbers):
    """
    Find the minimum number in a list.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The smallest number in the list.
    """
    return min(numbers)

print("1. Find Sum")
print("2. Find Average")
print("3. Find Maximum")
print("4. Find Minimum")

try:
    choice = int(input("Enter your choice (1-4): "))

    numbers_input = input("Enter numbers separated by spaces: ")

    if not numbers_input.strip():
        raise ValueError("Empty list is not allowed.")

    numbers = [float(num) for num in numbers_input.split()]

    if choice == 1:
        result = find_sum(numbers)
        print("Sum:", result)

    elif choice == 2:
        result = find_average(numbers)
        print("Average:", result)

    elif choice == 3:
        result = find_maximum(numbers)
        print("Maximum:", result)

    elif choice == 4:
        result = find_minimum(numbers)
        print("Minimum:", result)

    else:
        print("Invalid operation selected.")

except ValueError:
    print("Error: Please enter valid numeric values and a non-empty list.")
```

1. Flatten a Nested List:
Write a Python function that takes a nested list and flattens it into a single list, where all the elements
are in a single dimension.

• Requirements:

– Define a function flatten(lst) that takes a nested list lst and returns a flattened version
of the list.

– Example: For the input [[1, 2], [3, 4], [5]], the output should be [1, 2, 3, 4, 5].


```python
def flatten(lst):
    """
    Flatten a nested list into a single list (one level).

    Parameters:
    lst (list): Nested list.

    Returns:
    list: Flattened list.
    """
    result = []
    for sub in lst:
        for item in sub:
            result.append(item)
    return result

lst = [[1, 2], [3, 4], [5]]
print("Flatten:", flatten(lst))
```

3. Sum of All Elements in a Nested List:
Write a Python function that calculates the sum of all the numbers in a nested list (regardless of depth).

• Requirements:

– Define a function sum
nested(lst) that takes a nested list lst and returns the sum of all
the elements.

– Example: For the input [[1, 2], [3, [4, 5]], 6], the output should be 21.


```python
def sum_nested(lst):
    """
    Calculate the sum of all numbers in a nested list (any depth).

    Parameters:
    lst (list): Nested list.

    Returns:
    int/float: Sum of all elements.
    """
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)
        else:
            total += item
    return total

lst = [[1,2],[3,[4,5]],6]
print("Sum nested:", sum_nested(lst))
```

5. Find the Maximum Element in a Nested List:
Write a Python function that finds the maximum element in a nested list (regardless of depth).

• Requirements:

– Define a function find
element.
max(lst) that takes a nested list lst and returns the maximum

– Example: For the input [[1, 2], [3, [4, 5]], 6], the output should be 6.


```python
def find_max(lst):
    """
    Find the maximum element in a nested list.

    Parameters:
    lst (list): Nested list.

    Returns:
    int/float: Maximum value.
    """
    maximum = float('-inf')
    for item in lst:
        if isinstance(item, list):
            maximum = max(maximum, find_max(item))
        else:
            maximum = max(maximum, item)
    return maximum

lst = [[1,2],[3,[4,5]],6]
print("Max element:", find_max(lst))
```

```python
def count_occurrences(lst, elem):
    """
    Count occurrences of an element in a nested list.

    Parameters:
    lst (list): Nested list.
    elem : Element to count.

    Returns:
    int: Number of occurrences.
    """
    count = 0
    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)
        elif item == elem:
            count += 1
    return count

lst = [[1,2],[2,3],[2,4]]
print("Count occurrences:", count_occurrences(lst, 2))
```

7. Flatten a List of Lists of Lists:
Write a Python function that flattens a list of lists of lists into a single list, regardless of the depth.

• Requirements:

– Define a function deep
flatten(lst) that takes a deeply nested list lst and returns a single
f
lattened list.

– Example: For the input [[[1, 2], [3, 4]], [[5, 6], [7, 8]]], the output should be
[1, 2, 3, 4, 5, 6, 7, 8].


```python
def deep_flatten(lst):
    """
    Flatten a deeply nested list into a single list.

    Parameters:
    lst (list): Deeply nested list.

    Returns:
    list: Fully flattened list.
    """
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(deep_flatten(item))
        else:
            result.append(item)
    return result

lst = [[[1,2],[3,4]],[[5,6],[7,8]]]
print("Deep flatten:", deep_flatten(lst))
```

```python
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

# 1. Prove A * A^-1 = I
A_inv = np.linalg.inv(A)
print("\nA * A^-1:\n", np.dot(A, A_inv))

# 2. Show AB != BA
AB = np.dot(A,B)
BA = np.dot(B,A)

print("\nAB:\n", AB)
print("\nBA:\n", BA)

# 3. Prove (AB)^T = B^T A^T
left = np.transpose(np.dot(A,B))
right = np.dot(np.transpose(B), np.transpose(A))

print("\n(AB)^T:\n", left)
print("\nB^T A^T:\n", right)
```

sure and print the time taken for this operation.



```python
size = 1000

A = [[1]*size for _ in range(size)]
B = [[1]*size for _ in range(size)]

start = time.time()

result = [[sum(A[i][k] * B[k][j] for k in range(size))
           for j in range(size)]
           for i in range(size)]

end = time.time()

print("Python Matrix Multiplication Time:", end - start, "seconds")
```
